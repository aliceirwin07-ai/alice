```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from astropy import units as u
from astropy import constants as const
```

```python
df = pd.read_csv("../../course_assignments/homework9/solar_system.csv")
print(df.shape)
print(df.head())
print(df.tail())
print(df)
#transposing
df = df.set_index("Attribute").T
#setting the attributes to switch places with rows
df.index.name = "Planet"
#renaming stuff
df.reset_index(inplace=True)
#reset the indices that were messed up above
df.columns.name = None
#take away random names

print(df.shape) #10 rows 21 cols, the new shape is different because we got rid of one of the cols when we formatted it
print(df.loc[0])#16 have units, 4 don't
#mass,diameter,density,gravity,escape velocity,rotation period,length of day,distance from sun,perihelion,aphelion,
#orbital period,orbital velocity, orbital inclination,Obliquity to Orbit,mean temperature,surface pressure have units
#global magnetic field, ring system, number of moons, orbital eccentricity don't have units cuz they're just counting
print(df)
```

```python
print(df.dtypes)
for col in df:
    print(df[col].apply(type))
#all the types are strings
#it's a problem for data analysis because we can't do any math with strings such as adding strings together like we can with floats
```

```python
for col in df:
    if col!="Planet" and col!="Ring System?" and col!="Global Magnetic Field?":
        df[col] = pd.to_numeric(df[col], errors="coerce")
for col in df:
    print(df[col].apply(type))

#the 3 types of data are str - words and stuff,ints - whole numbers,floats - decimal numbers
#the cols are objects because we didn't change them
```

```python
def attach_units(col):
    units = u.km/u.s
    values = []
    for val in df[col]:
        values.append(val*units)
    df[col] = values
attach_units("Orbital Velocity (km/s)")
print(df["Orbital Velocity (km/s)"])
```

```python
semi_major_axis = (df["Perihelion (10^6 km)"] + df["Aphelion (10^6 km)"])/2
df.insert(11,"semi_major_axis (10^6 km)",semi_major_axis)
print(df.loc[0])
```

```python
newvalues = []
for val in df["Orbital Period (days)"]:
    newvalues.append((val/365)*u.year)
df["Orbital Period (days)"] = newvalues
df.rename(columns={"Orbital Period (days)": "Orbital Period (years)"}, inplace=True)
print(df.loc[0])
print(df["Orbital Period (years)"])
print(f"{df["Orbital Period (years)"][0]:.4f}")#mercury orbital period
```

```python
#AU is the avg length between us and the sun
print(const.au)
au_in_km = u.AU.to(u.km)
print(au_in_km)
```

```python
kmcols = ["Diameter (km)","Distance from Sun (10^6 km)","Perihelion (10^6 km)","Aphelion (10^6 km)","semi_major_axis (10^6 km)"]
for col in kmcols:
    newvalues = []
    if col != "Diameter (km)":
        for val in df[col]:
            # Convert from 10^6 km to AU
            newvalues.append(val*1e6*u.AU/(1.496*1e8))
        df[col] = newvalues
    else:
        for val in df[col]:
            # Convert from km to AU
            newvalues.append(val*u.AU/(1.496*1e8))
        df[col] = newvalues
df.rename(columns={"Diameter (km)": "Diameter (AU)"}, inplace=True)
df.rename(columns={"Distance from Sun (10^6 km)": "Distance from Sun (AU)"}, inplace=True)
df.rename(columns={"Perihelion (10^6 km)": "Perihelion (AU)"}, inplace=True)
df.rename(columns={"Aphelion (10^6 km)": "Aphelion (AU)"}, inplace=True)
df.rename(columns={"semi_major_axis (10^6 km)": "semi_major_axis (AU)"}, inplace=True)
print(df["Distance from Sun (AU)"][0])#mercury dist to the sun
```

```python
df.to_csv("units.csv", index=False)
```
